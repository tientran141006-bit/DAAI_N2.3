```python
import pandas as pd
import json

# Đọc dữ liệu thô và dữ liệu geography đã chuẩn hoá để check validate
df_customers_raw = pd.read_csv('customers.csv')
df_geo_3nf = pd.read_csv('geography_3nf.csv')

# --- 3. Tạo bảng CUSTOMERS_3NF ---
# Đảm bảo customer_id là duy nhất (PK)
df_cust_3nf = df_customers_raw.drop_duplicates(subset=['customer_id']).copy()

# Loại bỏ cột 'city' vì đã được chuẩn hóa sang GEOGRAPHY
if 'city' in df_cust_3nf.columns:
    df_cust_3nf = df_cust_3nf.drop(columns=['city'])

# --- 4. Kiểm tra Validation ---
valid_zips = set(df_geo_3nf['zip'])
cust_zips = set(df_cust_3nf['zip'])

invalid_fks = cust_zips - valid_zips
num_invalid_fks = len(invalid_fks)

# Báo cáo số liệu
report = {
    "customers": {
        "raw_rows": len(df_customers_raw),
        "3nf_rows": len(df_cust_3nf),
        "duplicates_removed": len(df_customers_raw) - len(df_cust_3nf),
        "invalid_fks": num_invalid_fks,
        "dropped_columns": ['city']
    }
}

# --- 5. Xuất ra file CSV ---
df_cust_3nf.to_csv('customer_3nf.csv', index=False)
print(json.dumps(report, indent=2))
```
