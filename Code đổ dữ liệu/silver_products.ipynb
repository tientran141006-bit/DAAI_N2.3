```python
import pandas as pd
import json

# Đọc dữ liệu thô
df_products_raw = pd.read_csv('products.csv')

# --- 2. Tạo bảng PRODUCTS_3NF ---
# Đảm bảo product_id là duy nhất (PK)
df_prod_3nf = df_products_raw.drop_duplicates(subset=['product_id']).copy()

# Báo cáo số liệu
report = {
    "products": {
        "raw_rows": len(df_products_raw),
        "3nf_rows": len(df_prod_3nf),
        "duplicates_removed": len(df_products_raw) - len(df_prod_3nf)
    }
}

# --- Xuất ra file CSV ---
df_prod_3nf.to_csv('product_3nf.csv', index=False)
print(json.dumps(report, indent=2))
```
