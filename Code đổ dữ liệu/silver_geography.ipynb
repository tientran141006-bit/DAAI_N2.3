```python
import pandas as pd
import json

# Đọc dữ liệu thô
df_geography_raw = pd.read_csv('geography.csv')

# --- 1. Tạo bảng GEOGRAPHY_3NF ---
# Đảm bảo zip là duy nhất (PK)
df_geo_3nf = df_geography_raw.drop_duplicates(subset=['zip']).copy()

# Báo cáo số liệu
report = {
    "geography": {
        "raw_rows": len(df_geography_raw),
        "3nf_rows": len(df_geo_3nf),
        "duplicates_removed": len(df_geography_raw) - len(df_geo_3nf)
    }
}

# --- Xuất ra file CSV ---
df_geo_3nf.to_csv('geography_3nf.csv', index=False)
print(json.dumps(report, indent=2))
```
